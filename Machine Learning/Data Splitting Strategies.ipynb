# The Standard Hold-Out
## Goal: Understand the impact of random seeds and basic splitting.
#### Task: Use train_test_split on the California Housing dataset.
#### Deliverable: Perform the split twice using different random_state values. Train a simple LinearRegression model on both and compare the Mean Squared Error (MSE) on the respective test sets. Observe how much the evaluation metric fluctuates just based on the split.

```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error

X, y = load_diabetes(return_X_y=True)

for seed in [42, 99]:
    X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.2, random_state=seed)
    model = LinearRegression().fit(X_tr, y_tr)
    mse = mean_squared_error(y_te, model.predict(X_te))
    print(f'seed={seed} | MSE={mse:.4f}')
```
