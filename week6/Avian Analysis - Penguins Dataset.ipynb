# Avian Analysis - Penguins Dataset
#### Dataset: sns.load_dataset('penguins') 
#### Goal: Differentiate penguin species by comparing multiple physical traits at once.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

penguins = sns.load_dataset('penguins')
penguins.head()
```

#### Adding Hue, Size, and Style: Build a sns.scatterplot() comparing bill_length_mm vs. bill_depth_mm. Make it 4-dimensional by assigning hue='species' (different colors for species), style='island' (different marker shapes for locations), and size='body_mass_g' (bubble sizes for weight).

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=penguins, 
    x='bill_length_mm', 
    y='bill_depth_mm', 
    hue='species', 
    style='island', 
    size='body_mass_g',
    sizes=(20, 200), 
    alpha=0.7        
)

plt.title('Penguin Bill Measurements by Species, Island, and Mass')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

English: This graph compares the bill length (x-axis) and bill depth (y-axis) of the penguins. By adding hue='species', we assign a unique color to each species, revealing how they naturally cluster based on their bill shape. The style='island' parameter changes the shape of the data points based on where the penguins live, and size='body_mass_g' changes the size of the bubble based on the penguin's weight. This allows us to effectively visualize four different variables at the exact same time in a single 2-dimensional plot.

#### Pair Plots: Run sns.pairplot(penguins, hue='species'). Have the students observe how this one line of code generates a massive grid of scatterplots and KDEs, instantly showing which variables easily separate the Adelie, Chinstrap, and Gentoo penguins.

```python
sns.pairplot(penguins, hue='species', markers=["o", "s", "D"])
plt.show()
```

English: The pair plot creates a massive grid of scatterplots for every numerical variable compared against every other numerical variable in the dataset. The diagonal line of graphs shows the distribution (Density/KDE plots) of each individual trait. Coloring by species (hue='species') instantly highlights which traits are best at separating the Adelie, Chinstrap, and Gentoo penguins. For example, you can easily see that Gentoo penguins are clearly separated from the others by their high body mass and long flippers.

Khmer: ក្រាហ្វ Pair plot នេះបង្កើតជាផ្ទាំងក្រឡានៃ scatterplots សម្រាប់អថេរជាលេខទាំងអស់ ប្រៀបធៀបគ្នាទៅវិញទៅមកនៅក្នុងសំណុំទិន្នន័យ។ បន្ទាត់កែងនៃក្រាហ្វ (diagonal) បង្ហាញពីរបាយទិន្នន័យ (Density/KDE plots) នៃលក្ខណៈនីមួយៗ។ ការដាក់ពណ៌តាមប្រភេទសត្វ (hue='species') បង្ហាញយ៉ាងច្បាស់ភ្លាមៗថាតើលក្ខណៈមួយណាដែលអាចបែងចែកសត្វភេនឃ្វីន Adelie, Chinstrap និង Gentoo បានល្អបំផុត។ ឧទាហរណ៍ អ្នកអាចមើលឃើញយ៉ាងងាយស្រួលថា សត្វភេនឃ្វីន Gentoo ត្រូវបែងចែកដាច់ពីគេយ៉ាងច្បាស់ដោយសារទម្ងន់ធ្ងន់ និងប្រវែងស្លាបវែងរបស់ពួកវា។


#### Correlation & Heatmaps: Have them drop the categorical columns, calculate the correlation matrix (df.corr()), and visualize it using sns.heatmap(annot=True, cmap='coolwarm'). They should discover the strong positive correlation between flipper length and body mass.

```python
numeric_penguins = penguins.select_dtypes(include=['float64', 'int64'])

correlation_matrix = numeric_penguins.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix, 
    annot=True,          
    cmap='coolwarm',     
    vmin=-1, vmax=1,     
    linewidths=0.5
)
plt.title('Correlation Matrix of Penguin Physical Traits')
plt.show()
```

English: A heatmap visualizes the correlation matrix, which tells us the statistical relationship between numerical variables. The values range from -1 to 1. A value close to 1 (red/warm colors) indicates a strong positive correlation (as one increases, the other increases). A value close to -1 (blue/cool colors) indicates a strong negative correlation (as one increases, the other decreases). Looking at the graph, we can discover a very strong positive correlation (usually around 0.87) between flipper_length_mm and body_mass_g, meaning that heavier penguins almost always have longer flippers.
