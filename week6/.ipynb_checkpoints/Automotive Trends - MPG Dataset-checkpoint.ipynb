# Automotive Trends - MPG Dataset\

#### Dataset: sns.load_dataset('mpg')
#### Goal: Compare car performance and fuel efficiency across different global regions.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

mpg = sns.load_dataset('mpg')

mpg.head()
```

#### FacetGrid / Relplot / Catplot: Use sns.relplot(x='weight', y='mpg', col='origin'). This will generate three separate scatterplots side-by-side (USA, Europe, Japan) so students can compare weight vs. fuel efficiency by region. Then, use sns.catplot(kind='box', x='cylinders', y='horsepower', row='origin') to create a vertical stack of boxplots.

```python
sns.relplot(
    data=mpg, 
    x='weight', 
    y='mpg', 
    col='origin', 
    hue='origin',
    height=4, 
    aspect=1
)

plt.subplots_adjust(top=0.85) 
plt.suptitle('Car Weight vs. MPG by Region of Origin')
plt.show()
```

English: This visualization uses sns.relplot to create a FacetGrid, generating three side-by-side scatterplots comparing a car's weight (x-axis) to its fuel efficiency in miles per gallon or MPG (y-axis). Each plot represents a different region of origin: USA, Japan, and Europe.

Observation: Across all regions, there is a clear negative correlation—as the car gets heavier, its fuel efficiency decreases. You can also easily see that cars from the USA tend to be much heavier and less fuel-efficient overall, while Japanese and European cars cluster toward the lighter, highly efficient end of the graph.

Khmer: ការបង្ហាញរូបភាពនេះប្រើប្រាស់ sns.relplot ដើម្បីបង្កើតជា FacetGrid ដែលបង្កើតបានជាក្រាហ្វចំណុច (Scatterplots) ចំនួន ៣ ទន្ទឹមគ្នា ដើម្បីប្រៀបធៀបទម្ងន់រថយន្ត (អ័ក្ស x) ជាមួយនឹងកម្រិតស៊ីប្រេងគិតជាម៉ាយល៍ក្នុងមួយហ្គាឡុង ឬ MPG (អ័ក្ស y)។ ក្រាហ្វនីមួយៗតំណាងឱ្យតំបន់ផលិតផ្សេងៗគ្នា៖ សហរដ្ឋអាមេរិក ជប៉ុន និងអឺរ៉ុប។

ការសង្កេត៖ នៅគ្រប់តំបន់ទាំងអស់ យើងឃើញមានទំនាក់ទំនងអវិជ្ជមានយ៉ាងច្បាស់លាស់ពោលគឺ ពេលរថយន្តកាន់តែធ្ងន់ កម្រិតសន្សំសំចៃប្រេងរបស់វាកាន់តែថយចុះ។ អ្នកក៏អាចមើលឃើញយ៉ាងងាយស្រួលផងដែរថា រថយន្តមកពីសហរដ្ឋអាមេរិកភាគច្រើនមានទម្ងន់ធ្ងន់ និងស៊ីប្រេងច្រើនជាងគេ ខណៈដែលរថយន្តជប៉ុន និងអឺរ៉ុបប្រមូលផ្តុំគ្នានៅផ្នែករថយន្តស្រាល និងស៊ីប្រេងតិច។

```python
sns.catplot(
    data=mpg, 
    kind='box', 
    x='cylinders', 
    y='horsepower', 
    row='origin', 
    hue='cylinders',
    palette='viridis',
    height=3, 
    aspect=2
)

plt.subplots_adjust(top=0.9)
plt.suptitle('Distribution of Horsepower by Cylinders Across Regions')
plt.show()
```

English: This visualization uses sns.catplot to create a vertical stack of boxplots. The boxplots show the distribution of horsepower (y-axis) based on the number of engine cylinders (x-axis), with a separate row for each region of origin. The boxes represent the middle 50% of the data (with a line for the median), while the "whiskers" show the range of typical values and the individual dots represent outliers.

Observation: You can see that more cylinders generally mean more horsepower. The graph clearly highlights that only the USA manufactured a significant number of 8-cylinder cars, which have the highest horsepower, whereas Europe and Japan primarily focused on 4-cylinder engines.

Khmer: ការបង្ហាញរូបភាពនេះប្រើប្រាស់ sns.catplot ដើម្បីបង្កើតជាក្រាហ្វប្រអប់ (Boxplots) តម្រៀបជាជួរឈរ។ ក្រាហ្វប្រអប់នេះបង្ហាញពីរបាយនៃកម្លាំងសេះ (អ័ក្ស y) ផ្អែកលើចំនួនស៊ីឡាំងម៉ាស៊ីន (អ័ក្ស x) ដោយបែងចែកជាជួរដេកដាច់ដោយឡែកសម្រាប់តំបន់ផលិតនីមួយៗ។ ប្រអប់នីមួយៗតំណាងឱ្យទិន្នន័យ ៥០% នៅចំកណ្តាល (មានបន្ទាត់សម្រាប់តម្លៃមេដ្យាន) ចំណែកឯ "បន្ទាត់សងខាង" បង្ហាញពីប្រវែងនៃតម្លៃទូទៅ ហើយចំណុចតូចៗតំណាងឱ្យតម្លៃខុសប្រក្រតី (Outliers)។

ការសង្កេត៖ អ្នកអាចមើលឃើញថាជាទូទៅ ស៊ីឡាំងកាន់តែច្រើន កម្លាំងសេះកាន់តែខ្លាំង។ ក្រាហ្វនេះបង្ហាញយ៉ាងច្បាស់ថា មានតែសហរដ្ឋអាមេរិកប៉ុណ្ណោះដែលផលិតរថយន្តប្រភេទ ៨ ស៊ីឡាំងច្រើនជាងគេ ដែលមានកម្លាំងសេះខ្លាំងបំផុត ចំណែកឯអឺរ៉ុប និងជប៉ុនផ្តោតសំខាន់លើម៉ាស៊ីន ៤ ស៊ីឡាំង។

#### Joint Plots: Use sns.jointplot(x='horsepower', y='mpg') to show a scatter plot in the center, with a histogram of horsepower on the top margin and a histogram of mpg on the right margin.

```python
sns.jointplot(
    data=mpg, 
    x='horsepower', 
    y='mpg', 
    kind='scatter', 
    color='purple',
    alpha=0.6       
)

plt.suptitle('Relationship and Distribution of Horsepower vs. MPG', y=1.02)
plt.show()
```

English: The sns.jointplot combines two different types of plots into one single figure. In the center, we have a scatter plot comparing horsepower (x-axis) and MPG (y-axis), which clearly shows that cars with higher horsepower get significantly lower gas mileage. On the outside margins, we have histograms (bar charts showing frequency).

Observation: The top histogram shows that most cars in this dataset have lower horsepower (clustering between 50 and 100), while the right margin histogram shows the overall distribution of MPG, peaking around the 15-20 and 25-30 MPG ranges.

Khmer: sns.jointplot រួមបញ្ចូលក្រាហ្វពីរប្រភេទផ្សេងគ្នាទៅក្នុងផ្ទាំងតែមួយ។ នៅចំកណ្តាល យើងមាន Scatter plot ដែលប្រៀបធៀបកម្លាំងសេះ (អ័ក្ស x) និងកម្រិតស៊ីប្រេង ឬ MPG (អ័ក្ស y) ដែលបង្ហាញយ៉ាងច្បាស់ថា រថយន្តដែលមានកម្លាំងសេះខ្លាំងជាង ស៊ីប្រេងច្រើនជាងយ៉ាងខ្លាំង។ នៅផ្នែកខាងក្រៅ យើងមាន Histogram (ក្រាហ្វសសរបង្ហាញពីភាពញឹកញាប់)។

ការសង្កេត៖ Histogram នៅផ្នែកខាងលើបង្ហាញថា រថយន្តភាគច្រើននៅក្នុងសំណុំទិន្នន័យនេះមានកម្លាំងសេះទាប (ប្រមូលផ្តុំគ្នាចន្លោះពី ៥០ ទៅ ១០០) ចំណែកឯ Histogram នៅខាងស្តាំបង្ហាញពីរបាយទូទៅនៃកម្រិតស៊ីប្រេង (MPG) ដែលមានចំនួនច្រើនបំផុតនៅចន្លោះ ១៥-២០ ម៉ាយល៍ និង ២៥-៣០ ម៉ាយល៍ក្នុងមួយហ្គាឡុង។
